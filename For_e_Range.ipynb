#FOR e Range

```python
nomes = ["Ana", "José", "Maria"]
for i in nomes:
  print("Olá, Seja Bem Vindo", i)
```

```python
for i in range(6):
  print(i)
```

```python
for i in range(1,6):
  print(i)
```

```python
for i in range(0,11,2):
  print(i)
```

```python
for i in range(10,0,-1):
  print(i)
```

```python
numeros = [4, 7, 10, 15, 22, 3]
for i in numeros:
  if i % 2 == 0:
    print(i, "é par")
  else:
    print(i, "é impar")
```

```python
numeros = range(11)
for i in numeros:
  if i % 2 == 0:
    print(i, "é par")
  else:
    print(i, "é impar")
```

```python
notas = [8.5, 9.2, 5.5, 3.0]
for nota in notas:
  if nota >= 6.0 and nota <=10.0:
    print(nota, "Aprovado")
  else:
    print(nota, "Reprovado")
```

```python
numeros = [4,7,10,15,22,3]
soma = 0
for numero in numeros:
  soma = soma + numero
print("A soma é: ", soma)
print("A média é: ", soma/len(numeros))
```

```python
for numero in range(1,11):
  if numero == 5:
    break
  print(numero)
```

```python
#6
notas = input("nota 1: "), input("nota2: "), input("nota 3: "), input("nota 4: "), input("nota 5: ")
somanotas = 0
for nt in notas:
  somanotas += (nt)

media = somanotas / len(notas)
print("A média das notas é:", media)
if media >= 7:
    print("A turma está com média boa!")
else:
    print("A turma precisa melhorar.")
```
